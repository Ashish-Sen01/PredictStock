# <span style="color:red"> SECTION 5 — Establish Strong Baselines </span>

```python
# VERIFY 7-DAY HISTORY AT FORECAST ORIGIN

history_start = FORECAST_ORIGIN - pd.Timedelta(days=6)

history_7d = daily_panel[
    (daily_panel["date"] >= history_start) &
    (daily_panel["date"] <= FORECAST_ORIGIN)
].copy()

history_pairs = (
    history_7d[["item_id", "store_id"]]
    .drop_duplicates()
)

history_counts = (
    history_7d
    .groupby(["item_id", "store_id"], observed=True)
    .size()
)

print("Required history dates:")
print(history_start, "→", FORECAST_ORIGIN)

print("\nHistory rows:", f"{len(history_7d):,}")
print("Unique pairs:", f"{len(history_pairs):,}")

print("\nExpected pairs:", len(forecast_pairs))
print("Pairs with exactly 7 days:", (history_counts == 7).sum())
print("Pairs with fewer than 7 days:", (history_counts < 7).sum())
print("Pairs with more than 7 days:", (history_counts > 7).sum())
```

```python
# PREPARE 7-DAY SEASONAL HISTORY

seasonal_history = (
    daily_panel[
        (daily_panel["date"] >= history_start) &
        (daily_panel["date"] <= FORECAST_ORIGIN)
    ][
        ["date", "item_id", "store_id", "quantity"]
    ]
    .copy()
)

seasonal_history["dow"] = seasonal_history["date"].dt.dayofweek

print("Seasonal history shape:", seasonal_history.shape)

print("\nDates:")
print(
    seasonal_history["date"].min(),
    "→",
    seasonal_history["date"].max()
)

print("\nUnique day-of-week values:",
      sorted(seasonal_history["dow"].unique()))
```

```python
# CHECK SEASONAL HISTORY COMPLETENESS

history_check = (
    seasonal_history
    .groupby(["item_id", "store_id"], observed=True)
    .agg(
        days=("date", "nunique"),
        unique_dow=("dow", "nunique")
    )
)

print("Pairs with 7 days and 7 unique weekdays:",
      ((history_check["days"] == 7) &
       (history_check["unique_dow"] == 7)).sum())

print("Total forecast pairs:",
      len(forecast_pairs))

print("\nIncomplete pairs:")
print(
    history_check[
        (history_check["days"] != 7) |
        (history_check["unique_dow"] != 7)
    ].head(10)
)
```

**# What the results mean**
* The required 7-day history is correctly identified:    
     2024-03-25 → 2024-03-31
* 13,300 / 13,447 forecast-origin pairs have a complete 7-day history.
* 147 pairs are incomplete.
* The 25,200 unique pairs in the history window are larger than the 13,447 forecast-origin universe because the history contains pairs that were not present on 2024-03-31.
* The incomplete pairs have fewer than 7 rows, so they should not be assigned a fake 7-day seasonal history.

* For our baseline, this means:   
We can use the 7-day seasonal baseline directly for the 13,300 complete pairs, while the 147 incomplete forecast pairs need a fallback method.

```python
# BUILD 7-DAY SEASONAL BASELINE

# Keep only forecast-origin pairs
seasonal_history_eval = seasonal_history.merge(
    forecast_pairs,
    on=["item_id", "store_id"],
    how="inner"
)

# Create a lookup using the last observed value for each pair + weekday
seasonal_lookup = (
    seasonal_history_eval
    .sort_values("date")
    .groupby(
        ["item_id", "store_id", "dow"],
        observed=True
    )["quantity"]
    .last()
    .reset_index()
    .rename(columns={"quantity": "baseline_7d"})
)

print("Seasonal lookup shape:", seasonal_lookup.shape)

print(
    "Expected rows for complete pairs:",
    13300 * 7
)

print(
    "Pairs represented:",
    seasonal_lookup[["item_id", "store_id"]]
    .drop_duplicates()
    .shape[0]
)
```

```python
# CREATE VALIDATION 7-DAY SEASONAL BASELINE

validation_baseline_7d = validation_eval.copy()

validation_baseline_7d["dow"] = (
    validation_baseline_7d["date"].dt.dayofweek
)

validation_baseline_7d = validation_baseline_7d.merge(
    seasonal_lookup,
    on=["item_id", "store_id", "dow"],
    how="left"
)

print("Validation baseline shape:",
      validation_baseline_7d.shape)

print(
    "Missing baseline predictions:",
    validation_baseline_7d["baseline_7d"].isna().sum()
)

print(
    "Total validation rows:",
    len(validation_baseline_7d)
)
```

```python
# CHECK BASELINE COVERAGE

missing_baseline = (
    validation_baseline_7d["baseline_7d"].isna()
)

missing_pairs = (
    validation_baseline_7d.loc[
        missing_baseline,
        ["item_id", "store_id"]
    ]
    .drop_duplicates()
)

print("Rows without 7-day baseline:",
      missing_baseline.sum())

print("Pairs without 7-day baseline:",
      len(missing_pairs))

print("\nExpected incomplete forecast pairs:",
      len(forecast_pairs) - 13300)

print("\nSample missing pairs:")
print(missing_pairs.head(10))
```

```python
# FILL INCOMPLETE 7-DAY BASELINE WITH HISTORICAL MEAN

# Mean demand available in the 7-day history
history_mean_lookup = (
    seasonal_history_eval
    .groupby(
        ["item_id", "store_id"],
        observed=True
    )["quantity"]
    .mean()
    .reset_index()
    .rename(columns={"quantity": "history_mean"})
)

validation_baseline_7d = validation_baseline_7d.merge(
    history_mean_lookup,
    on=["item_id", "store_id"],
    how="left"
)

# Use 7-day seasonal value where available,
# otherwise use the available-history mean.
validation_baseline_7d["baseline_7d_final"] = (
    validation_baseline_7d["baseline_7d"]
    .fillna(validation_baseline_7d["history_mean"])
)

print(
    "Missing final baseline predictions:",
    validation_baseline_7d["baseline_7d_final"].isna().sum()
)

print(
    "Fallback rows used:",
    validation_baseline_7d["baseline_7d"].isna().sum()
    if "baseline_7d" not in validation_baseline_7d.columns
    else validation_baseline_7d["baseline_7d"].isna().sum()
)
```

```python
# VERIFY FINAL BASELINE VALUES

print("Baseline statistics:")
print(
    validation_baseline_7d["baseline_7d_final"].describe()
)

print("\nNegative baseline values:",
      (validation_baseline_7d["baseline_7d_final"] < 0).sum())

print("\nZero baseline predictions:",
      (validation_baseline_7d["baseline_7d_final"] == 0).sum())

print("\nMissing baseline predictions:",
      validation_baseline_7d["baseline_7d_final"].isna().sum())
```

```python
# VERIFY FALLBACK PAIRS

fallback_pairs = (
    validation_baseline_7d.loc[
        validation_baseline_7d["baseline_7d"].isna(),
        ["item_id", "store_id", "history_mean"]
    ]
    .drop_duplicates()
)

print("Fallback pairs:", len(fallback_pairs))

print("\nFallback pair sample:")
print(fallback_pairs.head(10))

print("\nFallback mean statistics:")
print(fallback_pairs["history_mean"].describe())
```

```python
# EVALUATE 7-DAY SEASONAL BASELINE

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_true = validation_baseline_7d["actual_quantity"]
y_pred = validation_baseline_7d["baseline_7d_final"]

mae_7d = mean_absolute_error(y_true, y_pred)
rmse_7d = np.sqrt(mean_squared_error(y_true, y_pred))
r2_7d = r2_score(y_true, y_pred)

actual_total = y_true.sum()
pred_total = y_pred.sum()
bias_total = pred_total - actual_total
bias_pct = (bias_total / actual_total) * 100

print("7-Day Seasonal Baseline — Validation")
print("-" * 45)
print(f"MAE          : {mae_7d:.4f}")
print(f"RMSE         : {rmse_7d:.4f}")
print(f"R²           : {r2_7d:.4f}")
print(f"Actual total : {actual_total:,.2f}")
print(f"Predicted    : {pred_total:,.2f}")
print(f"Bias         : {bias_total:,.2f}")
print(f"Bias %       : {bias_pct:.2f}%")
```

```python
# ANALYZE BASELINE ERROR

validation_baseline_7d["error"] = (
    validation_baseline_7d["baseline_7d_final"]
    - validation_baseline_7d["actual_quantity"]
)

validation_baseline_7d["absolute_error"] = (
    validation_baseline_7d["error"].abs()
)

print("Error statistics:")
print(
    validation_baseline_7d["error"].describe()
)

print("\nAbsolute error statistics:")
print(
    validation_baseline_7d["absolute_error"].describe()
)

print(
    "\nRows with exact prediction:",
    (validation_baseline_7d["error"] == 0).sum()
)

print(
    "Rows over-predicted:",
    (validation_baseline_7d["error"] > 0).sum()
)

print(
    "Rows under-predicted:",
    (validation_baseline_7d["error"] < 0).sum()
)
```

```python
# 7-DAY BASELINE PERFORMANCE BY STORE

store_baseline_metrics = []

for store in sorted(
    validation_baseline_7d["store_id"].unique()
):
    subset = validation_baseline_7d[
        validation_baseline_7d["store_id"] == store
    ]

    actual = subset["actual_quantity"]
    pred = subset["baseline_7d_final"]

    store_baseline_metrics.append({
        "store_id": store,
        "rows": len(subset),
        "MAE": mean_absolute_error(actual, pred),
        "RMSE": np.sqrt(mean_squared_error(actual, pred)),
        "R2": r2_score(actual, pred),
        "actual_total": actual.sum(),
        "predicted_total": pred.sum(),
        "bias_pct": (
            (pred.sum() - actual.sum())
            / actual.sum()
        ) * 100
    })

store_baseline_metrics = pd.DataFrame(
    store_baseline_metrics
)

print(store_baseline_metrics.to_string(index=False))
```

* Exact predictions: 298,334 / 1,210,230 ≈ 24.65%
* Over-predictions: 519,585 ≈ 42.93%
* Under-predictions: 392,311 ≈ 32.42%
* Median error = 0, which is encouraging for a sparse-demand dataset.
* The very large maximum absolute error (1,010) shows that a relatively small number of observations have substantial errors.

* A second strong baseline will be useful: a recent-demand baseline based on the last 28 days. That gives us a different reference from the 7-day seasonal pattern

```python
# PREPARE 28-DAY RECENT-DEMAND BASELINE

recent_history_start = FORECAST_ORIGIN - pd.Timedelta(days=27)

recent_history = daily_panel[
    (daily_panel["date"] >= recent_history_start) &
    (daily_panel["date"] <= FORECAST_ORIGIN)
][
    ["date", "item_id", "store_id", "quantity"]
].copy()

print("Recent history:")
print(
    recent_history["date"].min(),
    "→",
    recent_history["date"].max()
)

print("\nRows:", f"{len(recent_history):,}")

print(
    "Unique forecast pairs represented:",
    recent_history[
        ["item_id", "store_id"]
    ].drop_duplicates().shape[0]
)

print(
    "Expected forecast pairs:",
    len(forecast_pairs)
)
```

```python
# CALCULATE 28-DAY MEAN DEMAND

recent_mean_lookup = (
    recent_history
    .groupby(
        ["item_id", "store_id"],
        observed=True
    )["quantity"]
    .mean()
    .reset_index()
    .rename(columns={"quantity": "baseline_28d"})
)

print("28-day mean lookup shape:",
      recent_mean_lookup.shape)

print("\nBaseline statistics:")
print(
    recent_mean_lookup["baseline_28d"].describe()
)

print(
    "\nMissing pair means:",
    recent_mean_lookup["baseline_28d"].isna().sum()
)
```

```python
# CREATE VALIDATION 28-DAY BASELINE

validation_baseline_28d = validation_eval.merge(
    recent_mean_lookup,
    on=["item_id", "store_id"],
    how="left"
)

print("Validation baseline shape:",
      validation_baseline_28d.shape)

print(
    "Missing baseline predictions:",
    validation_baseline_28d["baseline_28d"].isna().sum()
)

print(
    "Total validation rows:",
    len(validation_baseline_28d)
)
```

```python
# EVALUATE 28-DAY RECENT-DEMAND BASELINE

y_true_28 = validation_baseline_28d["actual_quantity"]
y_pred_28 = validation_baseline_28d["baseline_28d"]

mae_28d = mean_absolute_error(y_true_28, y_pred_28)
rmse_28d = np.sqrt(mean_squared_error(y_true_28, y_pred_28))
r2_28d = r2_score(y_true_28, y_pred_28)

actual_total_28 = y_true_28.sum()
pred_total_28 = y_pred_28.sum()
bias_total_28 = pred_total_28 - actual_total_28
bias_pct_28 = (bias_total_28 / actual_total_28) * 100

print("28-Day Recent-Mean Baseline — Validation")
print("-" * 48)
print(f"MAE          : {mae_28d:.4f}")
print(f"RMSE         : {rmse_28d:.4f}")
print(f"R²           : {r2_28d:.4f}")
print(f"Actual total : {actual_total_28:,.2f}")
print(f"Predicted    : {pred_total_28:,.2f}")
print(f"Bias         : {bias_total_28:,.2f}")
print(f"Bias %       : {bias_pct_28:.2f}%")
```

**We'll build the next approach around the evidence we've collected:**
* Stage 1: Determine whether demand is likely to be zero/non-zero

```python
# CREATE HYBRID BASELINE

validation_hybrid = validation_baseline_7d[
    ["date", "item_id", "store_id", "actual_quantity", "baseline_7d_final"]
].copy()

validation_hybrid = validation_hybrid.merge(
    validation_baseline_28d[
        ["date", "item_id", "store_id", "baseline_28d"]
    ],
    on=["date", "item_id", "store_id"],
    how="inner"
)

# Equal-weight combination
validation_hybrid["baseline_hybrid"] = (
    0.5 * validation_hybrid["baseline_7d_final"] +
    0.5 * validation_hybrid["baseline_28d"]
)

print("Hybrid baseline shape:",
      validation_hybrid.shape)

print(
    "Missing hybrid predictions:",
    validation_hybrid["baseline_hybrid"].isna().sum()
)

print("\nHybrid statistics:")
print(
    validation_hybrid["baseline_hybrid"].describe()
)
```

## <span style="color:red"> SECTION 7 — DEMAND STABILITY & REGIME-SHIFT DIAGNOSTICS </span>

```python
# Create Demand Stability Categories

stability_analysis = activity_compare.merge(
    magnitude_compare[
        ["item_id", "store_id", "mean_change_ratio"]
    ],
    on=["item_id", "store_id"],
    how="inner"
)

def classify_stability(row):
    activity_change = row["activity_change"]
    magnitude_ratio = row["mean_change_ratio"]

    if activity_change >= 0.25 and magnitude_ratio >= 1.25:
        return "Strong Increase"
    elif activity_change <= -0.25 and magnitude_ratio <= 0.75:
        return "Strong Decrease"
    elif activity_change >= 0.15 and magnitude_ratio >= 1.10:
        return "Moderate Increase"
    elif activity_change <= -0.15 and magnitude_ratio <= 0.90:
        return "Moderate Decrease"
    else:
        return "Relatively Stable"

stability_analysis["stability_class"] = stability_analysis.apply(
    classify_stability,
    axis=1
)

print("Stability analysis shape:", stability_analysis.shape)

print("\nStability categories:")
print(
    stability_analysis["stability_class"]
    .value_counts()
)

print("\nPercentage:")
print(
    stability_analysis["stability_class"]
    .value_counts(normalize=True).mul(100).round(2)
)
```

## <span style="color:red"> SECTION 8 — RECENT-WINDOW BASELINE ROBUSTNESS </span>

```python
# Prepare Recent-Windows

WINDOWS = [14, 56]

recent_window_history = daily_panel[
    (daily_panel["date"] > FORECAST_ORIGIN - pd.Timedelta(days=max(WINDOWS)))
    & (daily_panel["date"] <= FORECAST_ORIGIN)
].copy()

print("Recent-window history:")
print("Date range:", recent_window_history["date"].min(), "→",
      recent_window_history["date"].max())
print("Rows:", len(recent_window_history))
print("Unique pairs:",
      recent_window_history[["item_id", "store_id"]].drop_duplicates().shape[0])
```

```python
# Build 14-Day and 56-Day Baselines

window_lookups = {}

for window in WINDOWS:
    start_date = FORECAST_ORIGIN - pd.Timedelta(days=window - 1)

    temp = recent_window_history[
        recent_window_history["date"] >= start_date
    ]

    lookup = (
        temp.groupby(["item_id", "store_id"], observed=True)["quantity"]
        .mean()
        .reset_index()
        .rename(columns={"quantity": f"baseline_{window}d"})
    )

    window_lookups[window] = lookup

    print(
        f"{window}-day lookup:",
        lookup.shape,
        "| date range:",
        temp["date"].min(),
        "→",
        temp["date"].max()
    )
```

```python
# Create Validation Predictions

validation_window_baselines = validation_eval.copy()

for window in WINDOWS:
    lookup = window_lookups[window]

    validation_window_baselines = validation_window_baselines.merge(
        lookup,
        on=["item_id", "store_id"],
        how="left"
    )

    # Fallback to the 28-day baseline if the window has no value
    validation_window_baselines[f"baseline_{window}d"] = (
        validation_window_baselines[f"baseline_{window}d"]
        .fillna(validation_window_baselines["baseline_28d"]
                if "baseline_28d" in validation_window_baselines.columns
                else 0)
    )

print("Shape:", validation_window_baselines.shape)

for window in WINDOWS:
    col = f"baseline_{window}d"
    print(
        f"{col}: missing =",
        validation_window_baselines[col].isna().sum()
    )
```

```python
# Rebuild Complete Validation Baseline Panel
# Start from the complete validation evaluation panel
validation_window_baselines = validation_eval.copy()

# Add the already-established 7-day baseline
validation_window_baselines = validation_window_baselines.merge(
    validation_baseline_7d[
        ["date", "item_id", "store_id", "baseline_7d_final"]
    ],
    on=["date", "item_id", "store_id"],
    how="left"
)

# Add the already-established 28-day baseline
validation_window_baselines = validation_window_baselines.merge(
    validation_baseline_28d[
        ["date", "item_id", "store_id", "baseline_28d"]
    ],
    on=["date", "item_id", "store_id"],
    how="left"
)

# Add 14-day and 56-day baselines
for window in [14, 56]:
    lookup = window_lookups[window]

    validation_window_baselines = validation_window_baselines.merge(
        lookup,
        on=["item_id", "store_id"],
        how="left"
    )

    # Fallback to 28-day baseline if needed
    validation_window_baselines[f"baseline_{window}d"] = (
        validation_window_baselines[f"baseline_{window}d"]
        .fillna(validation_window_baselines["baseline_28d"])
    )

print("Shape:", validation_window_baselines.shape)

print("\nColumns:")
print(validation_window_baselines.columns.tolist())

print("\nMissing values:")
print(
    validation_window_baselines[
        [
            "actual_quantity",
            "baseline_7d_final",
            "baseline_14d",
            "baseline_28d",
            "baseline_56d"
        ]
    ].isna().sum()
)
```

```python
# Evaluate All Recent Windows
baseline_results = []

for window in [7, 14, 28, 56]:
    if window == 7:
        pred_col = "baseline_7d_final"
    elif window == 28:
        pred_col = "baseline_28d"
    else:
        pred_col = f"baseline_{window}d"

    actual = validation_window_baselines["actual_quantity"].to_numpy()
    pred = validation_window_baselines[pred_col].to_numpy()

    mae = np.mean(np.abs(actual - pred))
    rmse = np.sqrt(np.mean((actual - pred) ** 2))

    ss_res = np.sum((actual - pred) ** 2)
    ss_tot = np.sum((actual - actual.mean()) ** 2)
    r2 = 1 - (ss_res / ss_tot)

    actual_total = actual.sum()
    predicted_total = pred.sum()

    bias = predicted_total - actual_total
    bias_pct = bias / actual_total * 100

    baseline_results.append({
        "window": f"{window}D",
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Bias": bias,
        "Bias_%": bias_pct
    })

baseline_results_df = pd.DataFrame(baseline_results)

print(baseline_results_df.to_string(index=False))
```

## <span style="color:red"> SECTION 9 — VALIDATION VS TEST DISTRIBUTION SHIFT </span>

  * **4,889 pairs** showed increased demand.
  * **1,133 pairs** became inactive.
  * Only **15 pairs** became newly active.

* The **7,220 decreasing pairs** contributed about **933K units of demand loss**, while the increasing pairs recovered about **504K units**, resulting in the net decline within the original forecasting universe.

* Therefore, the previous **+93.49% XGBoost test overprediction cannot be explained by an overall demand collapse**.

* The diagnostics indicate that **product-store-level demand redistribution and changes in the forecasting universe are important factors** that the next forecasting approach must handle.

### Final Section 9 takeaway

> **Demand is relatively stable at the overall dataset level, but substantial demand redistribution occurs across product-store pairs. The forecasting challenge is therefore primarily one of pair-level demand variability and changing activity, rather than a simple global demand trend.**

```python
# Rebuild Test 56-Day Baseline with Zero-Filled History

# Create the complete 56-day history for every forecast pair
test_history_dates = pd.date_range(
    start=TEST_START - pd.Timedelta(days=56),
    end=TEST_START - pd.Timedelta(days=1),
    freq="D"
)

test_history_actual = sales_actual[
    (sales_actual["date"] >= TEST_START - pd.Timedelta(days=56)) &
    (sales_actual["date"] < TEST_START)
][["date", "item_id", "store_id", "quantity"]].copy()

# Create pair × date grid
test_history_grid = (
    forecast_pairs.assign(_key=1)
    .merge(
        pd.DataFrame({"date": test_history_dates, "_key": 1}),
        on="_key",
        how="inner"
    )
    .drop(columns="_key")
)

# Add actual demand
test_history_grid = test_history_grid.merge(
    test_history_actual,
    on=["date", "item_id", "store_id"],
    how="left"
)

# Missing sales dates = zero demand
test_history_grid["quantity"] = (
    test_history_grid["quantity"]
    .fillna(0)
)

# Calculate true 56-day calendar-day mean
test_56_lookup = (
    test_history_grid
    .groupby(["item_id", "store_id"], observed=True)["quantity"]
    .mean()
    .reset_index()
    .rename(columns={"quantity": "baseline_56d"})
)

# Merge with the complete test evaluation panel
test_baseline_56d = test_eval.merge(
    test_56_lookup,
    on=["item_id", "store_id"],
    how="left"
)

print("Test baseline shape:", test_baseline_56d.shape)

print("Missing predictions:",
      test_baseline_56d["baseline_56d"].isna().sum())

print("\nBaseline statistics:")
print(test_baseline_56d["baseline_56d"].describe())

print("\nActual total:",
      test_baseline_56d["actual_quantity"].sum())

print("Predicted total:",
      test_baseline_56d["baseline_56d"].sum())
```

```python
# Build 7-Day Direct Training Target

HORIZON = 7
origins_7d = training_origins[HORIZON]

# Keep only the rows needed for target construction
target_source_7d = daily_panel[
    ["date", "item_id", "store_id", "quantity"]
].copy()

target_parts_7d = []

for origin in origins_7d:
    target_start = origin + pd.Timedelta(days=1)
    target_end = origin + pd.Timedelta(days=HORIZON)

    future_data = target_source_7d[
        (target_source_7d["date"] >= target_start) &
        (target_source_7d["date"] <= target_end)
    ]

    target_7d = (
        future_data
        .groupby(["item_id", "store_id"], observed=True)["quantity"]
        .sum()
        .reset_index()
    )

    target_7d["origin_date"] = origin
    target_parts_7d.append(target_7d)

direct_target_7d = pd.concat(
    target_parts_7d,
    ignore_index=True
)

direct_target_7d = direct_target_7d.rename(
    columns={"quantity": "target_7d"}
)

print("7-Day Direct Target")
print("-" * 45)

print("Shape:", direct_target_7d.shape)
print("Unique origins:", direct_target_7d["origin_date"].nunique())
print("Unique pairs:", direct_target_7d[["item_id", "store_id"]].drop_duplicates().shape[0])

print("\nTarget statistics:")
print(direct_target_7d["target_7d"].describe())

print("\nMissing targets:",
      direct_target_7d["target_7d"].isna().sum())

print("Zero targets:",
      (direct_target_7d["target_7d"] == 0).sum())

print("Positive targets:",
      (direct_target_7d["target_7d"] > 0).sum())
```

```python
# Verify 7-Day Target Coverage by Origin

target_7d_by_origin = (
    direct_target_7d
    .groupby("origin_date", observed=True)
    .agg(
        rows=("target_7d", "size"),
        unique_pairs=("item_id", "size"),
        total_demand=("target_7d", "sum"),
        zero_targets=("target_7d", lambda x: (x == 0).sum()),
        positive_targets=("target_7d", lambda x: (x > 0).sum())
    )
    .reset_index()
)

print("7-Day Target Coverage by Origin")
print("-" * 70)

print(target_7d_by_origin.to_string(index=False))

print("\nCoverage summary:")
print("Minimum pairs per origin:",
      target_7d_by_origin["unique_pairs"].min())

print("Maximum pairs per origin:",
      target_7d_by_origin["unique_pairs"].max())

print("Average pairs per origin:",
      target_7d_by_origin["unique_pairs"].mean())

print("\nOrigins with fewer than 20,000 pairs:",
      (target_7d_by_origin["unique_pairs"] < 20000).sum())
```

```python
# Build 7-Day Direct Target Aligned to Origin Pairs

HORIZON = 7
origins_7d = training_origins[HORIZON]

target_source_7d = daily_panel[
    ["date", "item_id", "store_id", "quantity"]
].copy()

target_parts_7d = []

for origin in origins_7d:

    # Product-store pairs available at this forecasting origin
    origin_pairs = (
        daily_panel[
            daily_panel["date"] == origin
        ][["item_id", "store_id"]]
        .drop_duplicates()
    )

    # Future 7-day period
    target_start = origin + pd.Timedelta(days=1)
    target_end = origin + pd.Timedelta(days=HORIZON)

    future_data = target_source_7d[
        (target_source_7d["date"] >= target_start) &
        (target_source_7d["date"] <= target_end)
    ]

    # Aggregate future demand
    future_demand = (
        future_data
        .groupby(
            ["item_id", "store_id"],
            observed=True
        )["quantity"]
        .sum()
        .reset_index()
        .rename(columns={"quantity": "target_7d"})
    )

    # Align future demand to pairs available at the origin
    target_7d = origin_pairs.merge(
        future_demand,
        on=["item_id", "store_id"],
        how="left"
    )

    # No future transaction = zero demand
    target_7d["target_7d"] = (
        target_7d["target_7d"]
        .fillna(0)
    )

    target_7d["origin_date"] = origin

    target_parts_7d.append(target_7d)

direct_target_7d = pd.concat(
    target_parts_7d,
    ignore_index=True
)

print("7-Day Direct Target")
print("-" * 45)

print("Shape:", direct_target_7d.shape)
print("Unique origins:", direct_target_7d["origin_date"].nunique())

print(
    "Unique origin-pair combinations:",
    direct_target_7d[
        ["origin_date", "item_id", "store_id"]
    ].drop_duplicates().shape[0]
)

print("\nTarget statistics:")
print(direct_target_7d["target_7d"].describe())

print("\nMissing targets:",
      direct_target_7d["target_7d"].isna().sum())

print("Zero targets:",
      (direct_target_7d["target_7d"] == 0).sum())

print("Positive targets:",
      (direct_target_7d["target_7d"] > 0).sum())
```

```python
# Build 7-Day Direct Training Features

HORIZON = 7
origins_7d = training_origins[HORIZON]

feature_columns_7d = [
    "date",
    "item_id",
    "store_id",
    "pair_code",
    "day_of_week",
    "day_of_month",
    "week_of_year",
    "month",
    "quarter",
    "year",
    "is_weekend",
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",
    "rolling_mean_7",
    "rolling_mean_14",
    "rolling_mean_28",
    "month_sin",
    "month_cos",
    "dow_sin",
    "dow_cos",
    "dept_name",
    "class_name",
    "subclass_name",
    "item_type",
    "division",
    "format",
    "city",
    "area"
]

training_features_7d = daily_panel[
    daily_panel["date"].isin(origins_7d)
][feature_columns_7d].copy()

training_features_7d = training_features_7d.rename(
    columns={"date": "origin_date"}
)

print("7-Day Direct Training Features")
print("-" * 50)

print("Shape:", training_features_7d.shape)
print("Unique origins:",
      training_features_7d["origin_date"].nunique())

print(
    "Unique origin-pair combinations:",
    training_features_7d[
        ["origin_date", "item_id", "store_id"]
    ].drop_duplicates().shape[0]
)

print("\nMissing values:")
print(
    training_features_7d.isna()
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
# Create 56-Day Historical Demand Features

HORIZON = 7
origins_7d = training_origins[HORIZON]

history_56d_parts = []

for origin in origins_7d:

    history_start = origin - pd.Timedelta(days=56)

    history_data = daily_panel[
        (daily_panel["date"] >= history_start) &
        (daily_panel["date"] <= origin)
    ][
        ["date", "item_id", "store_id", "quantity"]
    ].copy()

    # Aggregate historical demand for each product-store pair
    history_stats = (
        history_data
        .groupby(
            ["item_id", "store_id"],
            observed=True
        )["quantity"]
        .agg(
            history_56d_mean="mean",
            history_56d_sum="sum",
            history_56d_std="std",
            history_56d_max="max"
        )
        .reset_index()
    )

    history_stats["origin_date"] = origin

    history_56d_parts.append(history_stats)

history_features_56d = pd.concat(
    history_56d_parts,
    ignore_index=True
)

# Standard deviation is zero for pairs with only one observation
history_features_56d["history_56d_std"] = (
    history_features_56d["history_56d_std"]
    .fillna(0)
)

print("56-Day Historical Features")
print("-" * 50)

print("Shape:", history_features_56d.shape)

print(
    "Unique origin-pair combinations:",
    history_features_56d[
        ["origin_date", "item_id", "store_id"]
    ].drop_duplicates().shape[0]
)

print("\nMissing values:")
print(
    history_features_56d.isna()
    .sum()
)

print("\nFeature statistics:")
print(
    history_features_56d[
        [
            "history_56d_mean",
            "history_56d_sum",
            "history_56d_std",
            "history_56d_max"
        ]
    ].describe()
)
```

```python
# Merge 7-Day Features, 56-Day History & Target

training_7d = training_features_7d.merge(
    history_features_56d,
    on=["origin_date", "item_id", "store_id"],
    how="left",
    validate="one_to_one"
)

training_7d = training_7d.merge(
    direct_target_7d,
    on=["origin_date", "item_id", "store_id"],
    how="inner",
    validate="one_to_one"
)

print("7-Day Direct Training Dataset")
print("-" * 55)

print("Shape:", training_7d.shape)

print(
    "Unique origin-pair combinations:",
    training_7d[
        ["origin_date", "item_id", "store_id"]
    ].drop_duplicates().shape[0]
)

print("\nMissing values:")
missing_training = (
    training_7d.isna()
    .sum()
    .sort_values(ascending=False)
)

print(missing_training[missing_training > 0])

print("\nTarget statistics:")
print(training_7d["target_7d"].describe())

print("\nTarget total:",
      f"{training_7d['target_7d'].sum():,.2f}")
```

## <span style="color:red"> DIRECT 7-DAY FORECASTING: BASELINE-ANCHORED MODEL </span>

Instead of asking the model to predict the entire 7-day demand from scratch, we'll make the 56-day baseline the central reference point and examine whether the historical features can predict the adjustment needed.

This is a different formulation:

$$ \text{target adjustment} = \text{actual 7-day demand} - \text{56-day baseline × 7} $$

The baseline is a daily mean, while target_7d is a 7-day total, so we must multiply the baseline by 7.

```python
# Create Baseline-Anchored 7-Day Target

model_train_7d["baseline_56d_7day"] = (
    model_train_7d["history_56d_mean"] * 7
)

model_valid_7d["baseline_56d_7day"] = (
    model_valid_7d["history_56d_mean"] * 7
)

model_train_7d["adjustment_7d"] = (
    model_train_7d["target_7d"]
    - model_train_7d["baseline_56d_7day"]
)

model_valid_7d["adjustment_7d"] = (
    model_valid_7d["target_7d"]
    - model_valid_7d["baseline_56d_7day"]
)

print("Baseline-Anchored Target")
print("=" * 60)

print("\nTraining adjustment:")
print(model_train_7d["adjustment_7d"].describe())

print("\nValidation adjustment:")
print(model_valid_7d["adjustment_7d"].describe())

print("\nMean adjustment:")
print(
    "Training   :",
    f"{model_train_7d['adjustment_7d'].mean():.6f}"
)
print(
    "Validation :",
    f"{model_valid_7d['adjustment_7d'].mean():.6f}"
)

print("\nMissing values:")
print(
    "Training   :",
    model_train_7d["adjustment_7d"].isna().sum()
)
print(
    "Validation :",
    model_valid_7d["adjustment_7d"].isna().sum()
)
```

```python
# Prepare Baseline-Anchored Model Features

ANCHOR_FEATURES_7D = [
    "item_id",
    "store_id",
    "pair_code",
    "day_of_week",
    "month",
    "quarter",
    "is_weekend",
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",
    "rolling_mean_7",
    "rolling_mean_14",
    "rolling_mean_28",
    "history_56d_mean",
    "history_56d_sum",
    "history_56d_std",
    "history_56d_max",
    "dept_name",
    "class_name",
    "subclass_name",
    "item_type",
    "division",
    "format",
    "city",
    "area"
]

X_train_anchor_7d = model_train_7d[ANCHOR_FEATURES_7D].copy()
y_train_anchor_7d = model_train_7d["adjustment_7d"].copy()

X_valid_anchor_7d = model_valid_7d[ANCHOR_FEATURES_7D].copy()
y_valid_anchor_7d = model_valid_7d["adjustment_7d"].copy()

anchor_categorical_cols_7d = [
    "item_id",
    "dept_name",
    "class_name",
    "subclass_name",
    "item_type",
    "division",
    "format",
    "city"
]

anchor_categorical_indices_7d = [
    X_train_anchor_7d.columns.get_loc(col)
    for col in anchor_categorical_cols_7d
]

print("Baseline-Anchored Model Data")
print("=" * 60)

print("Training features   :", X_train_anchor_7d.shape)
print("Training target     :", y_train_anchor_7d.shape)
print("Validation features :", X_valid_anchor_7d.shape)
print("Validation target   :", y_valid_anchor_7d.shape)

print("\nCategorical indices:")
print(anchor_categorical_indices_7d)

print("\nMissing values:")
print(
    "Training   :",
    X_train_anchor_7d.isna().sum().sum()
)
print(
    "Validation :",
    X_valid_anchor_7d.isna().sum().sum()
)
```

```python
# Train CatBoost Baseline-Anchored Model

catboost_anchor_7d = CatBoostRegressor(
    loss_function="MAE",
    eval_metric="MAE",
    iterations=500,
    depth=7,
    learning_rate=0.05,
    l2_leaf_reg=5,
    random_seed=42,
    verbose=100,
    task_type="CPU"
)

print("Training Baseline-Anchored CatBoost...")
print("=" * 60)

catboost_anchor_7d.fit(
    X_train_anchor_7d,
    y_train_anchor_7d,
    cat_features=anchor_categorical_indices_7d,
    eval_set=(X_valid_anchor_7d, y_valid_anchor_7d),
    use_best_model=True
)

print("\nTraining completed.")
print(
    "Best iteration:",
    catboost_anchor_7d.get_best_iteration()
)

print(
    "Best validation MAE:",
    catboost_anchor_7d
    .get_best_score()["validation"]["MAE"]
)
```

```python
# Convert Adjustment to 7-Day Demand Forecast

anchor_adjustment_pred_7d = (
    catboost_anchor_7d.predict(X_valid_anchor_7d)
)

baseline_valid_56d_7day = (
    model_valid_7d["baseline_56d_7day"].to_numpy()
)

anchor_pred_7d = (
    baseline_valid_56d_7day
    + anchor_adjustment_pred_7d
)

# Demand cannot be negative
anchor_pred_7d = np.maximum(
    anchor_pred_7d,
    0
)

print("Baseline-Anchored Forecast")
print("=" * 60)

print("Prediction count:", len(anchor_pred_7d))

print("\nPrediction statistics:")
print(
    pd.Series(anchor_pred_7d).describe()
)

print("\nActual total    :",
      f"{y_valid_7d.sum():,.2f}")

print("Predicted total :",
      f"{anchor_pred_7d.sum():,.2f}")

print("\nMissing predictions:",
      np.isnan(anchor_pred_7d).sum())

print("Negative predictions:",
      (anchor_pred_7d < 0).sum())
```

```python
# Build Multi-Window Historical Means

WINDOWS_DIRECT = [7, 14, 28, 56]

multi_window_parts = []

print("Building historical demand windows...")
print("=" * 60)

for origin in sorted(model_train_7d["origin_date"].unique()):
    
    origin = pd.Timestamp(origin)
    
    origin_pairs = model_train_7d[
        model_train_7d["origin_date"] == origin
    ][["item_id", "store_id"]].copy()
    
    window_features = origin_pairs.copy()
    window_features["origin_date"] = origin
    
    for window in WINDOWS_DIRECT:
        
        history_start = origin - pd.Timedelta(days=window - 1)
        
        history = daily_panel[
            (daily_panel["date"] >= history_start) &
            (daily_panel["date"] <= origin)
        ][
            ["date", "item_id", "store_id", "quantity"]
        ]
        
        history_mean = (
            history
            .groupby(
                ["item_id", "store_id"],
                observed=True
            )["quantity"]
            .mean()
            .reset_index()
            .rename(
                columns={
                    "quantity": f"mean_{window}d"
                }
            )
        )
        
        window_features = window_features.merge(
            history_mean,
            on=["item_id", "store_id"],
            how="left"
        )
    
    multi_window_parts.append(window_features)

multi_window_train = pd.concat(
    multi_window_parts,
    ignore_index=True
)

print("Training multi-window features:", multi_window_train.shape)
print("Missing values:")
print(multi_window_train.isna().sum())
```

```python
# Build Validation Multi-Window Means

multi_window_parts_valid = []

print("Building validation historical demand windows...")
print("=" * 60)

for origin in sorted(model_valid_7d["origin_date"].unique()):
    
    origin = pd.Timestamp(origin)
    
    origin_pairs = model_valid_7d[
        model_valid_7d["origin_date"] == origin
    ][["item_id", "store_id"]].copy()
    
    window_features = origin_pairs.copy()
    window_features["origin_date"] = origin
    
    for window in WINDOWS_DIRECT:
        
        history_start = origin - pd.Timedelta(days=window - 1)
        
        history = daily_panel[
            (daily_panel["date"] >= history_start) &
            (daily_panel["date"] <= origin)
        ][
            ["date", "item_id", "store_id", "quantity"]
        ]
        
        history_mean = (
            history
            .groupby(
                ["item_id", "store_id"],
                observed=True
            )["quantity"]
            .mean()
            .reset_index()
            .rename(
                columns={
                    "quantity": f"mean_{window}d"
                }
            )
        )
        
        window_features = window_features.merge(
            history_mean,
            on=["item_id", "store_id"],
            how="left"
        )
    
    multi_window_parts_valid.append(window_features)

multi_window_valid = pd.concat(
    multi_window_parts_valid,
    ignore_index=True
)

print("Validation multi-window features:", multi_window_valid.shape)
print("Missing values:")
print(multi_window_valid.isna().sum())
```

```python
# Evaluate Simple Window Blends

# Merge validation target
blend_eval = multi_window_valid.merge(
    model_valid_7d[
        ["origin_date", "item_id", "store_id", "target_7d"]
    ],
    on=["origin_date", "item_id", "store_id"],
    how="inner",
    validate="one_to_one"
)

y_blend = blend_eval["target_7d"]

blend_results = []

# Individual windows
for window in [7, 14, 28, 56]:
    
    pred = (
        blend_eval[f"mean_{window}d"] * 7
    )
    
    blend_results.append({
        "Model": f"{window}-Day Mean",
        "MAE": mean_absolute_error(y_blend, pred),
        "RMSE": np.sqrt(mean_squared_error(y_blend, pred)),
        "R2": r2_score(y_blend, pred),
        "Bias_%": (
            (pred.sum() - y_blend.sum())
            / y_blend.sum()
        ) * 100
    })

# Weighted combinations
blend_weights = {
    "50% 7D + 50% 56D":
        (0.50, 0.00, 0.00, 0.50),
    "25% 7D + 25% 28D + 50% 56D":
        (0.25, 0.00, 0.25, 0.50),
    "20% 7D + 20% 14D + 20% 28D + 40% 56D":
        (0.20, 0.20, 0.20, 0.40)
}

for name, (w7, w14, w28, w56) in blend_weights.items():
    
    pred = 7 * (
        w7 * blend_eval["mean_7d"]
        + w14 * blend_eval["mean_14d"]
        + w28 * blend_eval["mean_28d"]
        + w56 * blend_eval["mean_56d"]
    )
    
    blend_results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_blend, pred),
        "RMSE": np.sqrt(mean_squared_error(y_blend, pred)),
        "R2": r2_score(y_blend, pred),
        "Bias_%": (
            (pred.sum() - y_blend.sum())
            / y_blend.sum()
        ) * 100
    })

blend_results_df = pd.DataFrame(blend_results)

print("Historical Window & Blend Comparison")
print("=" * 100)
print(
    blend_results_df
    .sort_values("MAE")
    .to_string(index=False)
)
```

```python
# ============================================================
# CELL 123 — 30-DAY TRAINING ORIGINS
# ============================================================

HORIZON_30D = 30
TARGET_30D = "target_30d"

# training_origins is stored by horizon
training_origins_30d = pd.Series(
    pd.to_datetime(training_origins[30])
)

print("=" * 60)
print("30-DAY TRAINING ORIGINS")
print("=" * 60)

print("Number of origins :", len(training_origins_30d))
print("First origin      :", training_origins_30d.min())
print("Last origin       :", training_origins_30d.max())

print("\nExpected:")
print("~75 weekly origins")
print("2022-09-27 → 2024-02-27")
```

```python
# ============================================================
# CELL 124 — BUILD 30-DAY DIRECT TARGET
# ============================================================

# Create a lookup table for the established 30-day origins
origin_lookup_30d = pd.DataFrame({
    "origin_date": training_origins_30d
})

# Select only dates needed for the 30-day target windows
target_source_30d = daily_panel[
    daily_panel["date"].between(
        training_origins_30d.min() + pd.Timedelta(days=1),
        training_origins_30d.max() + pd.Timedelta(days=30)
    )
][["date", "item_id", "store_id", "quantity"]].copy()

# Assign each date to the most recent weekly origin
target_source_30d["origin_date"] = (
    target_source_30d["date"] -
    pd.to_timedelta(
        ((target_source_30d["date"] -
          training_origins_30d.min()).dt.days // 7) * 7,
        unit="D"
    )
)

# Keep only dates belonging to an established origin
target_source_30d = target_source_30d.merge(
    origin_lookup_30d,
    on="origin_date",
    how="inner"
)

# Keep the next 30 days after each origin
target_source_30d["days_ahead"] = (
    target_source_30d["date"] -
    target_source_30d["origin_date"]
).dt.days

target_source_30d = target_source_30d[
    target_source_30d["days_ahead"].between(1, 30)
]

direct_target_30d = (
    target_source_30d
    .groupby(
        ["origin_date", "item_id", "store_id"],
        observed=True
    )["quantity"]
    .sum()
    .reset_index()
    .rename(columns={"quantity": TARGET_30D})
)

print("=" * 60)
print("30-DAY DIRECT TARGET")
print("=" * 60)

print("Shape          :", direct_target_30d.shape)
print("Origins        :", direct_target_30d["origin_date"].nunique())
print("Target total   :", f"{direct_target_30d[TARGET_30D].sum():,.2f}")
print("Missing values :", direct_target_30d[TARGET_30D].isna().sum())
```

```python
# ============================================================
# CELL 124 — CORRECT 30-DAY DIRECT TARGET
# ============================================================

# Build the target separately for each established weekly origin.
# Target = total demand from origin+1 through origin+30.

target_parts_30d = []

for i, origin in enumerate(training_origins_30d, start=1):

    start_date = origin + pd.Timedelta(days=1)
    end_date = origin + pd.Timedelta(days=30)

    temp = daily_panel[
        daily_panel["date"].between(start_date, end_date)
    ][["item_id", "store_id", "quantity"]]

    target_part = (
        temp
        .groupby(["item_id", "store_id"], observed=True)["quantity"]
        .sum()
        .reset_index()
    )

    target_part["origin_date"] = origin

    target_parts_30d.append(target_part)

    if i % 10 == 0 or i == len(training_origins_30d):
        print(f"Processed {i}/{len(training_origins_30d)} origins")

direct_target_30d = pd.concat(
    target_parts_30d,
    ignore_index=True
)

direct_target_30d = direct_target_30d.rename(
    columns={"quantity": TARGET_30D}
)

direct_target_30d = direct_target_30d[
    ["origin_date", "item_id", "store_id", TARGET_30D]
]

print("\n" + "=" * 60)
print("30-DAY DIRECT TARGET — CORRECTED")
print("=" * 60)

print("Shape          :", direct_target_30d.shape)
print("Origins        :", direct_target_30d["origin_date"].nunique())
print("First origin   :", direct_target_30d["origin_date"].min())
print("Last origin    :", direct_target_30d["origin_date"].max())
print("Target total   :", f"{direct_target_30d[TARGET_30D].sum():,.2f}")
print("Missing values :", direct_target_30d[TARGET_30D].isna().sum())
```

```python
# ============================================================
# CELL 126 — BUILD 30-DAY TRAINING FEATURES
# ============================================================

# Historical feature window:
# Use the same 56-day history concept used in the 7-day model.

history_features_30d = (
    daily_panel[
        daily_panel["date"].isin(
            training_origins_30d
        )
    ]
    .copy()
)

# For each origin, calculate 56-day historical statistics
history_parts_30d = []

for i, origin in enumerate(training_origins_30d, start=1):

    start_date = origin - pd.Timedelta(days=55)
    end_date = origin

    temp = daily_panel[
        daily_panel["date"].between(start_date, end_date)
    ][["item_id", "store_id", "quantity"]]

    hist = (
        temp
        .groupby(["item_id", "store_id"], observed=True)["quantity"]
        .agg(
            history_56d_mean="mean",
            history_56d_std="std",
            history_56d_max="max"
        )
        .reset_index()
    )

    hist["origin_date"] = origin
    hist["history_56d_std"] = hist["history_56d_std"].fillna(0)

    history_parts_30d.append(hist)

    if i % 10 == 0 or i == len(training_origins_30d):
        print(f"Processed {i}/{len(training_origins_30d)} origins")

history_features_30d = pd.concat(
    history_parts_30d,
    ignore_index=True
)

print("\n" + "=" * 60)
print("30-DAY HISTORY FEATURES")
print("=" * 60)
print("Shape :", history_features_30d.shape)
print("Missing values:")
print(history_features_30d.isna().sum().sum())
```

```python
# ============================================================
# CELL 127 — MERGE 30-DAY FEATURES + TARGET
# ============================================================

# Use one row per origin × item × store from the daily panel.
# The latest available row at each origin provides calendar,
# lag and categorical features.

training_features_30d = daily_panel[
    daily_panel["date"].isin(training_origins_30d)
].copy()

training_features_30d = training_features_30d.rename(
    columns={"date": "origin_date"}
)

training_features_30d = training_features_30d.merge(
    history_features_30d,
    on=["origin_date", "item_id", "store_id"],
    how="left"
)

training_30d = training_features_30d.merge(
    direct_target_30d,
    on=["origin_date", "item_id", "store_id"],
    how="inner"
)

# Keep only the same 25 core features used for the 7-day benchmark.
DIRECT_FEATURES_30D = [
    "item_id",
    "store_id",
    "pair_code",
    "day_of_week",
    "month",
    "quarter",
    "is_weekend",
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",
    "rolling_mean_7",
    "rolling_mean_14",
    "rolling_mean_28",
    "history_56d_mean",
    "history_56d_std",
    "history_56d_max",
    "dept_name",
    "class_name",
    "subclass_name",
    "item_type",
    "division",
    "format",
    "city",
    "area"
]

print("=" * 60)
print("30-DAY TRAINING DATA")
print("=" * 60)
print("Shape :", training_30d.shape)
print("Features available :", len(DIRECT_FEATURES_30D))
print("Target :", TARGET_30D)
print("Missing values in features:",
      training_30d[DIRECT_FEATURES_30D].isna().sum().sum())
print("Missing targets:",
      training_30d[TARGET_30D].isna().sum())
```

## <span style="color:red">SECTION 12 — 90-DAY MODEL BENCHMARK </span>

```python
HORIZON_90D = 90
TARGET_90D = "target_90d"

training_origins_90d = pd.Series(
    pd.to_datetime(training_origins[90])
)

print("=" * 60)
print("90-DAY TRAINING ORIGINS")
print("=" * 60)
print("Number of origins :", len(training_origins_90d))
print("First origin      :", training_origins_90d.min())
print("Last origin       :", training_origins_90d.max())

# ------------------------------------------------------------
# Build 90-day target
# ------------------------------------------------------------

target_parts_90d = []

for i, origin in enumerate(training_origins_90d, start=1):

    start_date = origin + pd.Timedelta(days=1)
    end_date = origin + pd.Timedelta(days=90)

    temp = daily_panel[
        daily_panel["date"].between(start_date, end_date)
    ][["item_id", "store_id", "quantity"]]

    target_part = (
        temp
        .groupby(
            ["item_id", "store_id"],
            observed=True
        )["quantity"]
        .sum()
        .reset_index()
    )

    target_part["origin_date"] = origin

    target_parts_90d.append(target_part)

    if i % 10 == 0 or i == len(training_origins_90d):
        print(f"Processed {i}/{len(training_origins_90d)} origins")

direct_target_90d = pd.concat(
    target_parts_90d,
    ignore_index=True
)

direct_target_90d = direct_target_90d.rename(
    columns={"quantity": TARGET_90D}
)

direct_target_90d = direct_target_90d[
    ["origin_date", "item_id", "store_id", TARGET_90D]
]

print("\n" + "=" * 60)
print("90-DAY DIRECT TARGET")
print("=" * 60)
print("Shape          :", direct_target_90d.shape)
print("Origins        :", direct_target_90d["origin_date"].nunique())
print("Target total   :", f"{direct_target_90d[TARGET_90D].sum():,.2f}")
print("Missing values :", direct_target_90d[TARGET_90D].isna().sum())
```

```python
target_counts_90d = (
    direct_target_90d
    .groupby("origin_date", observed=True)
    .size()
)

print("=" * 60)
print("90-DAY TARGET VALIDATION")
print("=" * 60)

print("Unique origins :", len(target_counts_90d))
print("Minimum pairs  :", target_counts_90d.min())
print("Maximum pairs  :", target_counts_90d.max())
print("Mean pairs     :", target_counts_90d.mean())

print("\nTarget statistics:")
print(
    direct_target_90d[TARGET_90D].describe()
)

print(
    "\nZero-demand targets:",
    (direct_target_90d[TARGET_90D] == 0).sum()
)

print(
    "Positive targets:",
    (direct_target_90d[TARGET_90D] > 0).sum()
)

print(
    "Duplicate origin-item-store rows:",
    direct_target_90d.duplicated(
        ["origin_date", "item_id", "store_id"]
    ).sum()
)
```

```python
# ------------------------------------------------------------
# 90-DAY HISTORY FEATURES
# ------------------------------------------------------------

history_parts_90d = []

for i, origin in enumerate(training_origins_90d, start=1):

    start_date = origin - pd.Timedelta(days=55)
    end_date = origin

    temp = daily_panel[
        daily_panel["date"].between(start_date, end_date)
    ][["item_id", "store_id", "quantity"]]

    hist = (
        temp
        .groupby(
            ["item_id", "store_id"],
            observed=True
        )["quantity"]
        .agg(
            history_56d_mean="mean",
            history_56d_std="std",
            history_56d_max="max"
        )
        .reset_index()
    )

    hist["origin_date"] = origin
    hist["history_56d_std"] = hist["history_56d_std"].fillna(0)

    history_parts_90d.append(hist)

    if i % 10 == 0 or i == len(training_origins_90d):
        print(f"Processed {i}/{len(training_origins_90d)} origins")

history_features_90d = pd.concat(
    history_parts_90d,
    ignore_index=True
)

# ------------------------------------------------------------
# BUILD 90-DAY TRAINING DATA
# ------------------------------------------------------------

training_features_90d = daily_panel[
    daily_panel["date"].isin(training_origins_90d)
].copy()

training_features_90d = training_features_90d.rename(
    columns={"date": "origin_date"}
)

training_features_90d = training_features_90d.merge(
    history_features_90d,
    on=["origin_date", "item_id", "store_id"],
    how="left"
)

training_90d = training_features_90d.merge(
    direct_target_90d,
    on=["origin_date", "item_id", "store_id"],
    how="inner"
)

DIRECT_FEATURES_90D = [
    "item_id","store_id","pair_code","day_of_week","month","quarter",
    "is_weekend","lag_1","lag_7","lag_14","lag_28",
    "rolling_mean_7","rolling_mean_14","rolling_mean_28",
    "history_56d_mean","history_56d_std","history_56d_max",
    "dept_name","class_name","subclass_name","item_type","division",
    "format","city","area"
]

print("=" * 60)
print("90-DAY TRAINING DATA")
print("=" * 60)
print("Shape :", training_90d.shape)
print("Features :", len(DIRECT_FEATURES_90D))
print("Missing features :",
      training_90d[DIRECT_FEATURES_90D].isna().sum().sum())
print("Missing targets :",
      training_90d[TARGET_90D].isna().sum())
```

```python
# ------------------------------------------------------------
# CHRONOLOGICAL ORIGIN SPLIT
# ------------------------------------------------------------

unique_origins_90d = np.sort(
    training_90d["origin_date"].unique()
)

split_idx_90d = int(
    len(unique_origins_90d) * 0.80
)

train_origins_90d = unique_origins_90d[:split_idx_90d]
valid_origins_90d = unique_origins_90d[split_idx_90d:]

model_train_90d = training_90d[
    training_90d["origin_date"].isin(train_origins_90d)
].copy()

model_valid_90d = training_90d[
    training_90d["origin_date"].isin(valid_origins_90d)
].copy()

X_train_90d = model_train_90d[DIRECT_FEATURES_90D]
y_train_90d = model_train_90d[TARGET_90D]

X_valid_90d = model_valid_90d[DIRECT_FEATURES_90D]
y_valid_90d = model_valid_90d[TARGET_90D]

print("=" * 60)
print("90-DAY TRAIN / VALIDATION SPLIT")
print("=" * 60)

print("Total origins      :", len(unique_origins_90d))
print("Training origins   :", len(train_origins_90d))
print("Validation origins :", len(valid_origins_90d))

print("\nTraining period:")
print(train_origins_90d.min(), "→", train_origins_90d.max())

print("Validation period:")
print(valid_origins_90d.min(), "→", valid_origins_90d.max())

print("\nTraining shape   :", X_train_90d.shape)
print("Validation shape :", X_valid_90d.shape)

print("\nTraining target mean   :", y_train_90d.mean())
print("Validation target mean :", y_valid_90d.mean())

# ------------------------------------------------------------
# LIGHTGBM
# ------------------------------------------------------------

cat_features_90d = [
    "item_id","dept_name","class_name","subclass_name",
    "item_type","division","format","city"
]

lgbm_90d = LGBMRegressor(
    objective="regression_l1",
    n_estimators=600,
    learning_rate=0.05,
    num_leaves=63,
    max_depth=-1,
    min_child_samples=50,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=5,
    random_state=42,
    n_jobs=-1
)

lgbm_90d.fit(
    X_train_90d,
    y_train_90d,
    categorical_feature=cat_features_90d
)

pred_lgbm_90d = np.clip(
    lgbm_90d.predict(X_valid_90d),
    0,
    None
)

mae_lgbm_90d = mean_absolute_error(
    y_valid_90d,
    pred_lgbm_90d
)

rmse_lgbm_90d = np.sqrt(
    mean_squared_error(
        y_valid_90d,
        pred_lgbm_90d
    )
)

r2_lgbm_90d = r2_score(
    y_valid_90d,
    pred_lgbm_90d
)

actual_total_lgbm_90d = y_valid_90d.sum()
pred_total_lgbm_90d = pred_lgbm_90d.sum()

bias_lgbm_90d = (
    pred_total_lgbm_90d -
    actual_total_lgbm_90d
)

bias_pct_lgbm_90d = (
    bias_lgbm_90d /
    actual_total_lgbm_90d
) * 100

print("\n" + "=" * 60)
print("LIGHTGBM 90-DAY RESULTS")
print("=" * 60)

print(f"MAE       : {mae_lgbm_90d:.6f}")
print(f"RMSE      : {rmse_lgbm_90d:.6f}")
print(f"R²        : {r2_lgbm_90d:.6f}")
print(f"Actual    : {actual_total_lgbm_90d:,.2f}")
print(f"Predicted : {pred_total_lgbm_90d:,.2f}")
print(f"Bias      : {bias_lgbm_90d:,.2f}")
print(f"Bias %    : {bias_pct_lgbm_90d:.6f}%")
```

```python
# ============================================================
# SECTION 13 — FINAL MODEL PREPARATION
# Cell 1: Prepare full training datasets
# ============================================================

# Use all available historical origins for final training.
# The validation split is no longer used after model selection.

FINAL_FEATURES = [
    "item_id",
    "store_id",
    "pair_code",
    "day_of_week",
    "month",
    "quarter",
    "is_weekend",
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",
    "rolling_mean_7",
    "rolling_mean_14",
    "rolling_mean_28",
    "history_56d_mean",
    "history_56d_std",
    "history_56d_max",
    "dept_name",
    "class_name",
    "subclass_name",
    "item_type",
    "division",
    "format",
    "city",
    "area"
]

FINAL_CATEGORICAL = [
    "item_id",
    "dept_name",
    "class_name",
    "subclass_name",
    "item_type",
    "division",
    "format",
    "city"
]

# Full direct-training datasets
X_final_7d = training_7d[FINAL_FEATURES].copy()
y_final_7d = training_7d["target_7d"].copy()

X_final_30d = training_30d[FINAL_FEATURES].copy()
y_final_30d = training_30d["target_30d"].copy()

X_final_90d = training_90d[FINAL_FEATURES].copy()
y_final_90d = training_90d["target_90d"].copy()

print("FINAL TRAINING DATA")
print("=" * 70)

print(f"7-Day  : X={X_final_7d.shape}, y={y_final_7d.shape}")
print(f"30-Day : X={X_final_30d.shape}, y={y_final_30d.shape}")
print(f"90-Day : X={X_final_90d.shape}, y={y_final_90d.shape}")

print("\nTarget totals:")
print(f"7-Day  : {y_final_7d.sum():,.0f}")
print(f"30-Day : {y_final_30d.sum():,.0f}")
print(f"90-Day : {y_final_90d.sum():,.0f}")

print("\nMissing values:")
print("7-Day :", X_final_7d.isna().sum().sum())
print("30-Day:", X_final_30d.isna().sum().sum())
print("90-Day:", X_final_90d.isna().sum().sum())
```

```python
# ============================================================
# Cell 2: Train final LightGBM models
# ============================================================

from lightgbm import LGBMRegressor

final_model_lgbm_7d = LGBMRegressor(
    objective="regression_l1",
    n_estimators=600,
    learning_rate=0.05,
    num_leaves=63,
    max_depth=-1,
    min_child_samples=50,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=5,
    random_state=42,
    n_jobs=-1
)

final_model_lgbm_30d = LGBMRegressor(
    objective="regression_l1",
    n_estimators=600,
    learning_rate=0.05,
    num_leaves=63,
    max_depth=-1,
    min_child_samples=50,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=5,
    random_state=42,
    n_jobs=-1
)

final_model_lgbm_90d = LGBMRegressor(
    objective="regression_l1",
    n_estimators=600,
    learning_rate=0.05,
    num_leaves=63,
    max_depth=-1,
    min_child_samples=50,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=5,
    random_state=42,
    n_jobs=-1
)

final_model_lgbm_7d.fit(
    X_final_7d,
    y_final_7d,
    categorical_feature=FINAL_CATEGORICAL
)

final_model_lgbm_30d.fit(
    X_final_30d,
    y_final_30d,
    categorical_feature=FINAL_CATEGORICAL
)

final_model_lgbm_90d.fit(
    X_final_90d,
    y_final_90d,
    categorical_feature=FINAL_CATEGORICAL
)

print("FINAL LIGHTGBM MODELS TRAINED")
print("=" * 70)
print("7-Day  model : Ready")
print("30-Day model : Ready")
print("90-Day model : Ready")
```

```python
# ============================================================
# SECTION 14 — UNTOUCHED TEST FORECAST
# Cell 1 (CORRECTED): Build test-origin features
# ============================================================

TEST_ORIGIN = pd.Timestamp("2024-06-29")
HISTORY_START = TEST_ORIGIN - pd.Timedelta(days=55)

# Forecast universe from the original forecast origin
origin_pairs = (
    daily_panel.loc[
        daily_panel["date"] == pd.Timestamp("2024-03-31"),
        ["item_id", "store_id", "pair_code",
         "dept_name", "class_name", "subclass_name",
         "item_type", "division", "format", "city", "area"]
    ]
    .drop_duplicates(["item_id", "store_id"])
    .copy()
)

# IMPORTANT:
# Convert categorical IDs to their underlying values before creating
# the MultiIndex. This avoids the pandas Categorical/MultiIndex issue.
origin_pairs["item_id"] = origin_pairs["item_id"].astype(str)

history_sales = sales_actual[
    (sales_actual["date"] >= HISTORY_START) &
    (sales_actual["date"] <= TEST_ORIGIN)
][["date", "item_id", "store_id", "quantity"]].copy()

history_sales["date"] = pd.to_datetime(history_sales["date"])
history_sales["item_id"] = history_sales["item_id"].astype(str)

history_sales = (
    history_sales
    .groupby(["date", "item_id", "store_id"], as_index=False)["quantity"]
    .sum()
)

# Complete 56-day history
history_dates = pd.date_range(
    HISTORY_START,
    TEST_ORIGIN,
    freq="D"
)

# Use ordinary Index objects, not categorical indexes
item_ids = pd.Index(
    origin_pairs["item_id"].unique(),
    name="item_id"
)

store_ids = pd.Index(
    origin_pairs["store_id"].unique(),
    name="store_id"
)

# Create date × item × store grid
full_index = pd.MultiIndex.from_product(
    [history_dates, item_ids, store_ids],
    names=["date", "item_id", "store_id"]
)

history_grid = (
    history_sales
    .set_index(["date", "item_id", "store_id"])
    .reindex(full_index, fill_value=0)
    .reset_index()
)

# Keep only product-store combinations that actually belong
# to our forecast universe.
valid_pairs = origin_pairs[
    ["item_id", "store_id"]
].drop_duplicates()

history_grid = history_grid.merge(
    valid_pairs,
    on=["item_id", "store_id"],
    how="inner"
)

history_grid["quantity"] = history_grid["quantity"].astype("float32")

history_grid = history_grid.sort_values(
    ["item_id", "store_id", "date"]
)

group_cols = ["item_id", "store_id"]

# Lag features
history_grid["lag_1"] = (
    history_grid.groupby(group_cols)["quantity"].shift(1)
)

history_grid["lag_7"] = (
    history_grid.groupby(group_cols)["quantity"].shift(7)
)

history_grid["lag_14"] = (
    history_grid.groupby(group_cols)["quantity"].shift(14)
)

history_grid["lag_28"] = (
    history_grid.groupby(group_cols)["quantity"].shift(28)
)

# Rolling features
history_grid["rolling_mean_7"] = (
    history_grid.groupby(group_cols)["quantity"]
    .transform(lambda x: x.rolling(7, min_periods=7).mean())
)

history_grid["rolling_mean_14"] = (
    history_grid.groupby(group_cols)["quantity"]
    .transform(lambda x: x.rolling(14, min_periods=14).mean())
)

history_grid["rolling_mean_28"] = (
    history_grid.groupby(group_cols)["quantity"]
    .transform(lambda x: x.rolling(28, min_periods=28).mean())
)

# 56-day historical statistics
history_stats = (
    history_grid
    .groupby(group_cols)["quantity"]
    .agg(
        history_56d_mean="mean",
        history_56d_std="std",
        history_56d_max="max"
    )
    .reset_index()
)

# Forecast-origin row
test_features = history_grid[
    history_grid["date"] == TEST_ORIGIN
].copy()

test_features = test_features.merge(
    history_stats,
    on=group_cols,
    how="left"
)

# Remove metadata before merging the clean origin metadata
test_features = test_features.drop(
    columns=[
        "pair_code",
        "dept_name",
        "class_name",
        "subclass_name",
        "item_type",
        "division",
        "format",
        "city",
        "area"
    ],
    errors="ignore"
)

metadata_cols = [
    "item_id", "store_id", "pair_code",
    "dept_name", "class_name", "subclass_name",
    "item_type", "division", "format", "city", "area"
]

test_features = test_features.merge(
    origin_pairs[metadata_cols],
    on=["item_id", "store_id"],
    how="left"
)

# Calendar features at forecast origin
test_features["day_of_week"] = TEST_ORIGIN.dayofweek
test_features["month"] = TEST_ORIGIN.month
test_features["quarter"] = TEST_ORIGIN.quarter
test_features["is_weekend"] = int(TEST_ORIGIN.dayofweek >= 5)

# Match training categorical dtype
for col in FINAL_CATEGORICAL:
    if col in X_final_7d.columns and hasattr(X_final_7d[col], "cat"):
        test_features[col] = test_features[col].astype(
            X_final_7d[col].dtype
        )

print("TEST-ORIGIN FEATURES")
print("=" * 70)
print("Forecast origin :", TEST_ORIGIN.date())
print("History period  :", HISTORY_START.date(), "→", TEST_ORIGIN.date())
print("Rows            :", len(test_features))
print("Expected pairs  :", len(origin_pairs))
print("Columns         :", len(test_features.columns))

print("\nMissing model-feature values:")
print(
    test_features[FINAL_FEATURES]
    .isna()
    .sum()
    .sum()
)

print("\nFeature rows ready:",
      len(test_features) == len(origin_pairs))
```

```python
# ============================================================
# SECTION 14
# Cell 2: Generate untouched test forecasts
# ============================================================

X_test_origin = test_features[FINAL_FEATURES].copy()

# Match categorical dtypes used during training
for col in FINAL_CATEGORICAL:
    if hasattr(X_final_7d[col], "cat"):
        X_test_origin[col] = X_test_origin[col].astype(
            X_final_7d[col].dtype
        )

# Generate direct forecasts
test_pred_7d = final_model_lgbm_7d.predict(X_test_origin)
test_pred_30d = final_model_lgbm_30d.predict(X_test_origin)
test_pred_90d = final_model_lgbm_90d.predict(X_test_origin)

# Demand cannot be negative
test_pred_7d = np.maximum(test_pred_7d, 0)
test_pred_30d = np.maximum(test_pred_30d, 0)
test_pred_90d = np.maximum(test_pred_90d, 0)

# Store predictions
test_forecasts = test_features[
    ["item_id", "store_id", "pair_code"]
].copy()

test_forecasts["predicted_demand_7d"] = test_pred_7d
test_forecasts["predicted_demand_30d"] = test_pred_30d
test_forecasts["predicted_demand_90d"] = test_pred_90d

print("UNTOUCHED TEST FORECASTS GENERATED")
print("=" * 70)

print("Forecast origin :", TEST_ORIGIN.date())
print("Forecast pairs  :", len(test_forecasts))

print("\nTotal predicted demand:")
print(
    f"7-Day  : {test_forecasts['predicted_demand_7d'].sum():,.2f}"
)
print(
    f"30-Day : {test_forecasts['predicted_demand_30d'].sum():,.2f}"
)
print(
    f"90-Day : {test_forecasts['predicted_demand_90d'].sum():,.2f}"
)

print("\nPrediction statistics:")
print(
    test_forecasts[
        [
            "predicted_demand_7d",
            "predicted_demand_30d",
            "predicted_demand_90d"
        ]
    ].describe()
)
```

## <span style="color:red">SECTION 15 — UNTOUCHED TEST EVALUATION </span>

```python
# ============================================================
# SECTION 15 — UNTOUCHED TEST EVALUATION
# Cell 1: Build actual demand for test horizons
# ============================================================

TEST_START = pd.Timestamp("2024-06-30")
TEST_END = pd.Timestamp("2024-09-26")

# Forecast-pair universe
test_pairs = test_forecasts[
    ["item_id", "store_id"]
].drop_duplicates()

# Actual sales during untouched test period
test_actual_sales = sales_actual[
    (sales_actual["date"] >= TEST_START) &
    (sales_actual["date"] <= TEST_END)
][["date", "item_id", "store_id", "quantity"]].copy()

test_actual_sales["date"] = pd.to_datetime(test_actual_sales["date"])
test_actual_sales["item_id"] = test_actual_sales["item_id"].astype(str)

# Match pair ID type with forecast table
test_pairs = test_pairs.copy()
test_pairs["item_id"] = test_pairs["item_id"].astype(str)

# Keep ONLY the original forecast universe
test_actual_sales = test_actual_sales.merge(
    test_pairs,
    on=["item_id", "store_id"],
    how="inner"
)

# 7-day actual demand: Jun 30 -> Jul 6
actual_7d = (
    test_actual_sales[
        test_actual_sales["date"] <=
        TEST_START + pd.Timedelta(days=6)
    ]
    .groupby(["item_id", "store_id"])["quantity"]
    .sum()
    .rename("actual_demand_7d")
    .reset_index()
)

# 30-day actual demand: Jun 30 -> Jul 29
actual_30d = (
    test_actual_sales[
        test_actual_sales["date"] <=
        TEST_START + pd.Timedelta(days=29)
    ]
    .groupby(["item_id", "store_id"])["quantity"]
    .sum()
    .rename("actual_demand_30d")
    .reset_index()
)

# Available test period for the long horizon:
# Jun 30 -> Sep 26 = 89 days.
actual_89d = (
    test_actual_sales
    .groupby(["item_id", "store_id"])["quantity"]
    .sum()
    .rename("actual_demand_89d")
    .reset_index()
)

# Merge actuals onto every forecast pair.
test_evaluation = test_forecasts.merge(
    actual_7d,
    on=["item_id", "store_id"],
    how="left"
)

test_evaluation = test_evaluation.merge(
    actual_30d,
    on=["item_id", "store_id"],
    how="left"
)

test_evaluation = test_evaluation.merge(
    actual_89d,
    on=["item_id", "store_id"],
    how="left"
)

# Missing means zero demand.
actual_cols = [
    "actual_demand_7d",
    "actual_demand_30d",
    "actual_demand_89d"
]

test_evaluation[actual_cols] = (
    test_evaluation[actual_cols]
    .fillna(0)
)

print("TEST ACTUAL DEMAND CREATED")
print("=" * 70)

print("Test period :", TEST_START.date(), "→", TEST_END.date())
print("Test days   :", (TEST_END - TEST_START).days + 1)
print("Pairs       :", len(test_evaluation))

print("\nActual demand totals:")
print(
    f"7-Day  : {test_evaluation['actual_demand_7d'].sum():,.2f}"
)
print(
    f"30-Day : {test_evaluation['actual_demand_30d'].sum():,.2f}"
)
print(
    f"89-Day : {test_evaluation['actual_demand_89d'].sum():,.2f}"
)

print("\nMissing actual values:")
print(
    test_evaluation[actual_cols]
    .isna()
    .sum()
    .sum()
)
```

```python
# ============================================================
# Cell 2: Final untouched test metrics
# ============================================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def evaluate_test_forecast(actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual, predicted)

    actual_total = actual.sum()
    predicted_total = predicted.sum()

    bias = predicted_total - actual_total
    bias_pct = (
        (bias / actual_total) * 100
        if actual_total != 0 else np.nan
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Actual Total": actual_total,
        "Predicted Total": predicted_total,
        "Bias": bias,
        "Bias %": bias_pct
    }

results_7d_test = evaluate_test_forecast(
    test_evaluation["actual_demand_7d"],
    test_evaluation["predicted_demand_7d"]
)

results_30d_test = evaluate_test_forecast(
    test_evaluation["actual_demand_30d"],
    test_evaluation["predicted_demand_30d"]
)

print("FINAL UNTOUCHED TEST RESULTS")
print("=" * 70)

print("\n7-DAY DIRECT LIGHTGBM")
for k, v in results_7d_test.items():
    print(f"{k:18s}: {v:,.6f}" if isinstance(v, float) else f"{k:18s}: {v:,.2f}")

print("\n30-DAY DIRECT LIGHTGBM")
for k, v in results_30d_test.items():
    print(f"{k:18s}: {v:,.6f}" if isinstance(v, float) else f"{k:18s}: {v:,.2f}")

print("\n90-DAY MODEL")
print("90-day direct model cannot be fairly scored against")
print("the available 89-day test ground truth.")
print("Available actual window:", 
      test_evaluation["actual_demand_89d"].sum())
```

```python
# ============================================================
# SECTION 17
# Cell 1: Create final forecast output
# ============================================================

final_forecast = test_forecasts.copy()

# Keep clean column order
final_forecast = final_forecast[
    [
        "item_id",
        "store_id",
        "pair_code",
        "predicted_demand_7d",
        "predicted_demand_30d",
        "predicted_demand_90d"
    ]
].copy()

# Round forecasts for practical inventory use
final_forecast["predicted_demand_7d"] = (
    final_forecast["predicted_demand_7d"].clip(lower=0)
)

final_forecast["predicted_demand_30d"] = (
    final_forecast["predicted_demand_30d"].clip(lower=0)
)

final_forecast["predicted_demand_90d"] = (
    final_forecast["predicted_demand_90d"].clip(lower=0)
)

print("FINAL FORECAST TABLE")
print("=" * 70)

print("Rows   :", len(final_forecast))
print("Columns:", len(final_forecast.columns))

print("\nColumns:")
print(final_forecast.columns.tolist())

print("\nMissing values:")
print(final_forecast.isna().sum())

print("\nSample:")
display(final_forecast.head(10))
```
